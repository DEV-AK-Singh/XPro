# MATPLOTLIB

## 1. BAR PLOT 

```python
import matplotlib.pyplot as plt
x = ["java", "python", "C++"]
y2025 = [70, 90, 80]
y2026 = [80, 70, 90]
plt.title("Languages popularites")
plt.xlabel("Languages")
plt.ylabel("Popularty")
plt.bar(x, y2025, width=0.3, color="y", align="center", edgecolor="b", linewidth=2, linestyle=":", label="2025 trends")
plt.bar(x, y2026, width=0.3, color="r", align="center", edgecolor="b", linewidth=2, linestyle=":", label="2026 trends")
plt.legend()
plt.show()
```
